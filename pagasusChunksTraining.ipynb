```python
# Pegasus datasets
class PegasusDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item["labels"] = torch.tensor(
            self.labels["input_ids"][idx]
        )  # torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels["input_ids"])  # len(self.labels)
```

```python
def prepare_data(model_name, train_texts, train_labels,keywwordsList=[]):
    """
    Prepare input data for model fine-tuning
    """
    console.log(f"data preparation started with model {model_name}")
    tokenizer = PegasusTokenizer.from_pretrained(model_name)
    keywordsStr = " ".join(keywords)
    def tokenize_data(texts, labels,keyword_str):
        encodings = tokenizer(texts, truncation=True, padding=True)
        decodings = tokenizer(labels, truncation=True, padding=True)
        dataset_tokenized = PegasusDataset(encodings, decodings)
        return dataset_tokenized
    console.log(f"now tokenising the dataset of length = {len(train_texts)} with lables = {len(train_labels)}" )
    train_dataset = tokenize_data(train_texts, train_labels,keywordsStr)
    console.log(f"tokenizing done.... returning train_dataset and tokenizer.")
    return train_dataset, tokenizer
```

```python
def prepare_fine_tuning(model_name,tokenizer,train_dataset,val_dataset=None,freeze_encoder=False,output_dir="./results/",):
    console.log(f" fine tuning the {model_name} model.")
    torch_device = "cuda" if torch.cuda.is_available() else "cpu"
    print("torche device == ", torch_device)
    model = PegasusForConditionalGeneration.from_pretrained(model_name).to(torch_device)
    console.log(f" model captured = success.")
    training_args = TrainingArguments(
        output_dir=output_dir,
        num_train_epochs=3,
        per_device_train_batch_size=32,
	      per_device_eval_batch_size = 32,
        evaluation_strategy='steps',
        save_total_limit=1,
        eval_steps=1000,
        warmup_steps=500,
        weight_decay=0.01,
        logging_dir="./logs",
        logging_steps=150,
        save_strategy='steps',
        save_steps=500,
    )
    console.log(f"training arguments len = {len(training_args)}")
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        tokenizer=tokenizer,
    )
    console.log(f"trainer prepared = success!")
    return trainer,model

```

```python
if __name__ == "__main__":
    train_texts, train_labels = X_train, X_test
    # use Pegasus Large model as base for fine-tuning
    model_name = "google/pegasus-pubmed"
    console.log(f"model used:: {model_name}")
    train_dataset, tokenizer = prepare_data(model_name, train_texts, train_labels,keywordsList = keywords)
    console.log("data preparation done... ")
    trainer,model = prepare_fine_tuning(model_name, tokenizer, train_dataset)
    console.log("fine tuning done... starting training. ")
    trainer.train()
```
